### APPLY MODERN PORTFOLIO THEORY

```python
# Create an empty list for Portfolio returns, volatility and weights respectively
port_returns = []
port_volatility = []
port_weights = []

num_assets = len(df.columns)
num_portfs = 25

# Calculate the individual asset returns
individual_rets = df.resample('Y').last().pct_change().mean()
```

```python
for port in range(num_portfs):
    weights = np.random.random(num_assets) # This will randomly generate weight combinations
    weights = weights/np.sum(weights) # This will normalize weights so that they sum to 1
    port_weights.append(weights)
    returns = np.dot(weights, individual_rets)
    port_returns.append(returns)
    var = ret_cov.mul(weights, axis=0).mul(weights, axis=1).sum().sum() # This is to compute portfolio variance
    sd = np.sqrt(var)
    annual_sd = sd*np.sqrt(250) # Here we annualize the standard deviation to get the volatility
    port_volatility.append(annual_sd)
```

```python
# Flatten your column names to fix the MultiIndex columns issue, which usually occurs when you download multiple tickets from yfinance
close_df = df['Close']
print(close_df.columns)
```

```python
# Create a DataFrame so we can see the weight combinations for assets and corresponding returns and volatility
data = {'Returns': port_returns, 'Volatility': port_volatility}

for counter, symbol in enumerate(close_df.columns.tolist()):
    data[symbol +' weight'] = [w[counter] for w in port_weights]
```

```python
portfolio_V1 = pd.DataFrame(data)
portfolio_V1.head()
```

```python
# Plot Efficient Frontier
portfolio_V1.plot.scatter(x = 'Volatility', y = 'Returns', marker='o', color='y', s=15, alpha=0.5, grid=True, figsize=[8, 8])
plt.xlabel('Risk (Volatility)')
plt.ylabel('Expected Returns')
```
